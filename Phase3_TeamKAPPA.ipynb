#### <b>Data Cleaning</b>

The dataset for the Global Superstore is fairly clean, but after looking at the overall structure, there are a few minor data cleaning steps to take.
<ol>
<li>Identifiers: Dropping some columns like Row ID and Postal Code, which won't be useful for our analysis and modeling down the line. </li>
<li>Dates: The columns for Order Data and Ship Date are now converted to datetime. Another minor task would be to ensure that there is no Ship Date earlier than the Order Date.</li>
<li>Categorical Values: Minor normalization of spacing and casing as well as checking for spelling inconsistencies.</li></ol>

```python
import re

def clean_global_superstore(df: pd.DataFrame, *, verbose=True):
    """
    Cleans the Global Superstore dataframe with the following steps:
      1) Drop identifier-like columns that aren't useful for analysis
      2) Convert Order Date and Ship Date to datetime; flag any Ship Date earlier than Order Date.
      3) Normalize categorical values: trim spaces, collapse multiple spaces, and standardize casing.
    Returns: cleaned_df, report (dict with basic diagnostics)
    """
    df = df.copy()

    # 1) Drop low-value identifier columns if present
    drop_cols = [c for c in ["Row ID", "Postal Code"] if c in df.columns]
    df.drop(columns=drop_cols, inplace=True, errors="ignore")

    # 2) Dates to datetime 
    # Detect common name variants just in case
    order_date_col = next((c for c in df.columns if c.lower().strip().replace("_"," ") == "order date"), None)
    ship_date_col  = next((c for c in df.columns if c.lower().strip().replace("_"," ") == "ship date"), None)

    if order_date_col:
        df[order_date_col] = pd.to_datetime(df[order_date_col], errors="coerce")
    if ship_date_col:
        df[ship_date_col] = pd.to_datetime(df[ship_date_col], errors="coerce")

    # Flag rows where Ship Date is earlier than Order Date
    bad_ship_mask = None
    n_bad_ship = 0
    if order_date_col and ship_date_col:
        bad_ship_mask = (df[ship_date_col].notna() & df[order_date_col].notna() &
                         (df[ship_date_col] < df[order_date_col]))
        n_bad_ship = int(bad_ship_mask.sum())
        # Strategy: keep them but add a flag; you can decide to drop later if desired
        df["ship_before_order_flag"] = bad_ship_mask

    # 3) Categorical normalization (spacing + casing)
    # Only apply to common categorical columns; leave IDs/codes unchanged
    likely_cat_cols = [
        "Category", "Sub-Category", "Segment", "Ship Mode", "Region",
        "Country", "State", "City"
    ]
    cat_cols = [c for c in likely_cat_cols if c in df.columns]

    def _normalize_text(s: pd.Series) -> pd.Series:
        # strip, collapse internal whitespace, and Title Case
        s = s.astype("string")
        s = s.str.strip()
        s = s.str.replace(r"\s+", " ", regex=True)
        # Title-case but keep common acronyms from being mangled, if any appear (e.g., 'US')
        s = s.str.title()
        return s

    for c in cat_cols:
        df[c] = _normalize_text(df[c])

    # Basic spelling/inconsistency check report (case/space normalized vs original)
    spelling_report = {}
    for c in cat_cols:
        before = pd.Series(df[c].astype("string"))
        # simulate a lax normalization to spot near-duplicates
        normalized = (before.str.lower()
                             .str.replace(r"\s+", " ", regex=True)
                             .str.strip())
        # If original unique count is meaningfully higher than normalized, there were inconsistencies
        spelling_report[c] = {
            "unique_original": int(before.nunique(dropna=True)),
            "unique_normalized_lower": int(normalized.nunique(dropna=True))
        }

    report = {
        "dropped_columns": drop_cols,
        "date_columns": {"order_date_col": order_date_col, "ship_date_col": ship_date_col},
        "ship_before_order_count": n_bad_ship,
        "spelling_inconsistency_check": spelling_report
    }

    if verbose:
        print("Dropped columns:", drop_cols)
        if order_date_col or ship_date_col:
            print("Date columns:", report["date_columns"])
        if n_bad_ship:
            print(f"⚠️  Rows with Ship Date earlier than Order Date: {n_bad_ship} "
                  "(flagged in 'ship_before_order_flag')")
        else:
            print("No Ship Date < Order Date issues found (or date columns missing).")
        print("Categorical normalization applied to:", cat_cols)

    return df, report
```


#### <b>Visualize Outliers</b>
 Visualizes outliers before and after capping using boxplots for all numerical columns. Outliers were capped at a reasonable threshold (e.g., using IQR or percentile limits) to observe their behavior without removing them completely from the dataset.

```python
# Function to cap outliers using IQR
def cap_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return np.where(series > upper, upper, np.where(series < lower, lower, series))

# -----------------------------
# Identify numeric columns
num_cols = [
    col for col in df2.select_dtypes(include=[np.number]).columns
    if not any(keyword in col.lower() for keyword in ["id", "code"])
]
# Create capped dataframe
df2_capped = df2.copy()
for col in num_cols:
    df2_capped[col] = cap_outliers(df2[col])

# -----------------------------
# Plot boxplots before and after capping
plt.figure(figsize=(10, len(num_cols)*3))

for i, col in enumerate(num_cols, 1):
    # Before capping
    plt.subplot(len(num_cols), 2, 2*i-1)
    sns.boxplot(x=df2[col])
    plt.title(f'Original {col}')
    
    # After capping
    plt.subplot(len(num_cols), 2, 2*i)
    sns.boxplot(x=df2_capped[col])
    plt.title(f'Capped {col}')

plt.tight_layout()
plt.show()
```

Outliers were not removed in this analysis. Instead, they were handled using robust techniques such as log transformation, median/IQR-based statistics, and robust scaling to reduce their influence. This approach ensures that extreme values are retained for analysis (since they may represent important business cases like high-value customers) while minimizing their distortion on overall results.

#### <b>Data Cleaning</b>

The dataset for the E-Commerce Consumer Behavior has more data that needs to be handled before we proceed with further phases.
<ol>
<li>Identifiers: Indexes need to be trated as keys instead of features.</li>
<li>Dates and times: convert to datetime.</li>
<li>Categorical values: Normalize their spacing and formatting. </li>
<li>Nissing Values: Imputing a missing flag for values that are missing. </li>
<li>Numeric Value: Ensure that reordered is either 0 or 1 as well as validating non-negative counts. </li>
</ol>

```python
def clean_ecommerce_behavior(df: pd.DataFrame,*, verbose=True, build_proxy_datetime=False):
    """
    Cleaning the dataset for the E-Commerce following the steps written in the summary above.
    Returns
    -------
    cleaned_df : pd.DataFrame
    report     : dict
    """
    d = df.copy()

    # 1) Drop anonymous index cols & identify keys
    anon_cols = [c for c in d.columns if c.lower().startswith("unnamed")]
    d.drop(columns=anon_cols, inplace=True, errors="ignore")

    key_candidates = [c for c in ["order_id", "user_id", "product_id"] if c in d.columns]

    # 2) Dates / Time
    # Try to parse explicit timestamps if present
    ts_cols = [c for c in d.columns if any(k in c.lower() for k in ["timestamp", "purchase_time", "order_time", "order_date"])]
    parsed_ts = []
    for c in ts_cols:
        before_na = d[c].isna().sum()
        d[c] = pd.to_datetime(d[c], errors="coerce", infer_datetime_format=True)
        parsed_ts.append((c, before_na, d[c].isna().sum()))

    # Standardize dow and hour if present
    if "order_dow" in d.columns:
        d["order_dow"] = pd.to_numeric(d["order_dow"], errors="coerce")
        out_of_range_dow = (~d["order_dow"].between(0, 6)) & d["order_dow"].notna()
        if out_of_range_dow.any():
            d.loc[out_of_range_dow, "order_dow"] = np.nan

    if "order_hour_of_day" in d.columns:
        d["order_hour_of_day"] = pd.to_numeric(d["order_hour_of_day"], errors="coerce")
        out_of_range_hour = (~d["order_hour_of_day"].between(0, 23)) & d["order_hour_of_day"].notna()
        if out_of_range_hour.any():
            d.loc[out_of_range_hour, "order_hour_of_day"] = np.nan

    if build_proxy_datetime and ("order_dow" in d.columns) and ("order_hour_of_day" in d.columns):
        # Use an arbitrary base week (Monday=0) and add dow/hour offsets
        base = pd.Timestamp("2025-01-06") 
        dow_offset = pd.to_timedelta(d["order_dow"].fillna(0), unit="D")
        hour_offset = pd.to_timedelta(d["order_hour_of_day"].fillna(0), unit="h")
        d["order_proxy_ts"] = base + dow_offset + hour_offset

    # 3) Categorical normalization
    likely_text = [c for c in ["department", "aisle", "product_name"] if c in d.columns]
    def _norm_text(s: pd.Series) -> pd.Series:
        s = s.astype("string")
        s = s.str.strip()
        s = s.str.replace(r"\s+", " ", regex=True)
        s = s.str.lower()
        return s

    for c in likely_text:
        d[c] = _norm_text(d[c])

    # 4) Missing values
    if "days_since_prior_order" in d.columns:
        d["days_since_prior_order"] = pd.to_numeric(d["days_since_prior_order"], errors="coerce")
        d["days_since_prior_order_was_missing"] = d["days_since_prior_order"].isna()
        median_days = d["days_since_prior_order"].median(skipna=True)
        d["days_since_prior_order"] = d["days_since_prior_order"].fillna(median_days)

    # 5) Numeric sanity
    # reordered make sure to ensure {0,1}
    if "reordered" in d.columns:
        d["reordered"] = pd.to_numeric(d["reordered"], errors="coerce").fillna(0)
        # Anything not equal to 1 becomes 0
        d["reordered"] = (d["reordered"] == 1).astype(int)

    # price/quantity checks if present
    numeric_cols = d.select_dtypes(include=[np.number]).columns.tolist()
    for col in ["price", "quantity", "add_to_cart_order"]:
        if col in d.columns:
            d[col] = pd.to_numeric(d[col], errors="coerce")
            neg_mask = d[col] < 0
            if neg_mask.any():
                d[f"{col}_negative_flag"] = neg_mask

    # 6) Simple reports
    missing_pct = d.isna().mean().sort_values(ascending=False)
    n_rows = len(d)
    high_cardinality = [c for c in d.columns if d[c].nunique(dropna=True) > 0.9 * n_rows]

    # outof range trackers
    bad_hour_count = int((("order_hour_of_day" in d.columns) and d["order_hour_of_day"].isna().sum()) or 0)
    bad_dow_count = int((("order_dow" in d.columns) and d["order_dow"].isna().sum()) or 0)

    report = {
        "dropped_unnamed_columns": anon_cols,
        "keys_detected": key_candidates,
        "timestamp_columns_parsed": parsed_ts,
        "missing_pct_top": missing_pct.head(15),
        "high_cardinality_columns": high_cardinality,
        "bad_hour_entries_after_clamp_to_nan": bad_hour_count,
        "bad_dow_entries_after_clamp_to_nan": bad_dow_count,
    }

    if verbose:
        if anon_cols:
            print("Dropped unnamed index columns:", anon_cols)
        print("Keys detected:", key_candidates)
        if ts_cols:
            print("Parsed timestamp-like columns:", [c for c,_,_ in parsed_ts])
        if "days_since_prior_order" in d.columns:
            print("Imputed 'days_since_prior_order' with median and added flag 'days_since_prior_order_was_missing'.")
        if "reordered" in d.columns:
            print("Coerced 'reordered' to binary {0,1}.")
        if high_cardinality:
            print("High-cardinality columns (>90% unique):", high_cardinality)
        print("Top missing % columns:\n", missing_pct.head(10))

    return d, report
```

```python
# Normalize text helpers
def norm(s): 
    return s.astype("string").str.strip().str.replace(r"\s+"," ", regex=True).str.lower()

# ----- Global Superstore: sales by Category (top 8) -----
gs_cat = pd.Series(dtype=float)
if "Category" in df.columns and "Sales" in df.columns:
    gs_cat = (df.assign(_cat=norm(df["Category"]))
                .groupby("_cat")["Sales"]
                .sum()
                .sort_values(ascending=False)
                .head(8))

# ----- E-Commerce: reorder rate by department (top 8 by volume) -----
ec_dept = pd.Series(dtype=float)
if "department" in df2.columns:
    vol = df2.assign(_dept=norm(df2["department"]))
    # department share & reorder rate
    dept_stats = vol.groupby("_dept").agg(
        orders=("user_id","size"),
        reorder_rate=("reordered", lambda x: np.nan if x.isna().all() else x.mean())
    ).sort_values("orders", ascending=False).head(8)
    ec_dept = dept_stats["reorder_rate"]

# Plot side-by-side bars
fig, axes = plt.subplots(1, 2, figsize=(12,4), sharey=False)
if len(gs_cat):
    gs_cat.plot(kind="barh", ax=axes[0])
    axes[0].invert_yaxis()
    axes[0].set_title("GS: Sales by Category (Top 8)")
    axes[0].set_xlabel("Total Sales")
if len(ec_dept.dropna()):
    ec_dept.plot(kind="barh", ax=axes[1])
    axes[1].invert_yaxis()
    axes[1].set_title("EC: Reorder Rate by Department (Top 8 by Volume)")
    axes[1].set_xlabel("Reorder Rate")
plt.tight_layout(); plt.show()
```

Together, this highlights the contrast between what looks to be supply chain orders vs consumers.

This long-term view provides a macro-level understanding of revenue growth that complements the short-term behavioral patterns seen in the E-Commerce dataset.
